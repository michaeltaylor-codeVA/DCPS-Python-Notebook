## 1. A small sample with a known mean
The same ten prices from Lesson 2.4. Their mean is 528,480: check your script against it before trusting it on the full column.

```python
prices = [285000, 355000, 1111000, 1200000, 574900, 185000, 399900, 269000, 235000, 670000]
```

## 2. Running sum and count

```python
total = 0
count = 0
for price in prices:
    total = total + price
    count = count + 1
print(total, count)
```

```python
mean = total / count
print(mean)
```
