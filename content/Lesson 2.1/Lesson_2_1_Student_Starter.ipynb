## 3. Connect characters and Unicode code points

Unicode is a standardized character system. Each character has an assigned integer called a **code point**. Python's `ord()` function goes from a one-character string to its code point; `chr()` goes from a code point back to a character.

```python
ord('A')
```

```python
chr(65)
```

**Checkpoint:** Describe how `ord()` and `chr()` undo each other's work in the two cells above.

### Test characters of your choice

Replace `?` with **one** character. Run the cell at least three times with different choices. Try an uppercase letter, a lowercase letter, a digit, a punctuation mark, or an emoji.

```python
character = '?'
code_point = ord(character)
print("Character:", character)
print("Unicode code point:", code_point)
print("Back to a character:", chr(code_point))
```

## 4. Revisit binary

A code point is an integer, and integers can be written in binary. Run the cell to view the code point for `A` in base 10 and base 2.

```python
code_point = ord('A')
print("Base 10:", code_point)
print("Base 2: ", format(code_point, '08b'))
```

> **Precision note:** The Unicode code point is the character's assigned number. When text is saved or transmitted, an encoding such as UTF-8 converts code points into bytes. Today, we are exploring the code-point step of that process.
