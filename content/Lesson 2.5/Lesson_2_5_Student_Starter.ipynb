## 1. Load the raw dataset

```python
import pandas as pd
df = pd.read_csv('DC_Properties_300_Student.csv')
df.head()
```

```python
df.info()
```

## 3. Find missing values

```python
df.isnull().sum()
```

```python
df[['PRICE', 'SALEDATE']].isnull().sum()
```

## 4. Missing values pandas does not catch
Some rows have the letter `U` typed where a value should be. pandas reads it as text, so `.isnull()` does not count it.

```python
print((df['PRICE'] == 'U').sum())
```

```python
df[df['PRICE'] == 'U'][['SOURCE_ROW_ID', 'PRICE', 'SALEDATE', 'QUALIFIED']]
```

## 5. Why are prices missing? Check QUALIFIED

```python
df[df['PRICE'].isnull()]['QUALIFIED'].value_counts()
```

## 6. Is the reverse true? Does every U row have a missing price?

```python
print((df['QUALIFIED'] == 'U').sum())
```

```python
has_price = df['PRICE'].notnull() & (df['PRICE'] != 'U')
df[(df['QUALIFIED'] == 'U') & has_price][['SOURCE_ROW_ID', 'PRICE', 'QUALIFIED']].head()
```
