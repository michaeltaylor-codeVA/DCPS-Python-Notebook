## 1. Reload the raw dataset
This notebook starts fresh, so you do not need your Lesson 2.5 notebook.

````python
import pandas as pd
df = pd.read_csv('DC_Properties_300_Student.csv')
print(len(df))
````

## 2. Find and remove duplicate rows
Save the result in a new variable so the raw data is never lost.

````python
print(df.duplicated().sum())
````

````python
df_clean = df.drop_duplicates()
print(len(df_clean))
````

## 3. Handle missing PRICE values
First, turn the eight `U` placeholders from Lesson 2.5 into real missing values. Then drop every row with a missing price (our class decision).

````python
df_clean['PRICE'] = df_clean['PRICE'].replace('U', None)
print(df_clean['PRICE'].isnull().sum())
````

````python
df_clean = df_clean[df_clean['PRICE'].notnull()]
print(len(df_clean))
````

## 4. Convert PRICE to a number
PRICE is still text because of the commas (`285,000.00`). Remove the commas, then convert.

````python
df_clean['PRICE'] = pd.to_numeric(df_clean['PRICE'].str.replace(',', ''))
print(df_clean['PRICE'].dtype)
````

````python
print(round(df_clean['PRICE'].mean(), 2))
````

## 5. Save the cleaned dataset
The new file appears in this folder in the file browser. To keep a copy, right-click it and choose **Download**.

````python
df_clean.to_csv('dc_properties_clean.csv', index=False)
print(pd.read_csv('dc_properties_clean.csv').shape)
````
