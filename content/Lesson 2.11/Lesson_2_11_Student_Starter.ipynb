## Load the cleaned dataset
`dc_properties_clean.csv` is in this folder. It is the cleaned file from Lesson 2.6, provided here so you do not need your own copy.

```python
import pandas as pd
df_clean = pd.read_csv('dc_properties_clean.csv')
print(len(df_clean))
```

## 1. Rebuild the ward_counts dictionary from Lesson 2.7

```python
ward_counts = {}
for ward in df_clean['WARD']:
    if ward in ward_counts:
        ward_counts[ward] = ward_counts[ward] + 1
    else:
        ward_counts[ward] = 1
print(ward_counts)
```

## 2. A basic bar chart

```python
import matplotlib.pyplot as plt
plt.bar(ward_counts.keys(), ward_counts.values())
plt.show()
```

## 3. Customize the chart
`sorted()` puts the wards in order from Ward 1 to Ward 8.

```python
wards = sorted(ward_counts.keys())
counts = [ward_counts[w] for w in wards]
plt.bar(wards, counts, color='steelblue')
plt.title('Number of Properties by Ward')
plt.xlabel('Ward')
plt.ylabel('Number of Properties')
plt.xticks(rotation=45)
plt.show()
```

## 4. Save the chart as an image file
Build the chart, **save it before** `plt.show()`, then show it. (Saving after `plt.show()` produces a blank image.)

A pink warning that starts with `MatplotlibDeprecationWarning` may appear on this site. It is expected: the file still saves correctly.

The file appears in this folder in the file browser. To keep it, right-click it and choose **Download**.

```python
plt.bar(wards, counts, color='steelblue')
plt.title('Number of Properties by Ward')
plt.xlabel('Ward')
plt.ylabel('Number of Properties')
plt.xticks(rotation=45)
plt.savefig('ward_counts.png')
plt.show()
```
