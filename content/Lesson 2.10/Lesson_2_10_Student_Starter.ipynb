## 1. Import matplotlib and make a first plot

```python
import matplotlib.pyplot as plt
plt.plot([1, 2, 3], [4, 1, 9])
plt.show()
```

## Load the cleaned dataset
`dc_properties_clean.csv` is in this folder. It is the cleaned file from Lesson 2.6, provided here so you do not need your own copy.

```python
import pandas as pd
df_clean = pd.read_csv('dc_properties_clean.csv')
print(len(df_clean))
```

## 2. The bins from Lesson 1.10
Lesson 1.10 grouped PRICE into 21 bins, each $200,000 wide, from $0 up to $4,200,000. This list holds the bin edges.

```python
bins = list(range(0, 4200001, 200000))
print(len(bins) - 1, 'bins')
print(bins[:4], '...', bins[-2:])
```

## 3. The PRICE histogram

```python
plt.hist(df_clean['PRICE'], bins=bins)
plt.show()
```

## 4. Add a title and axis labels

```python
plt.hist(df_clean['PRICE'], bins=bins)
plt.title('Distribution of DC Property Prices')
plt.xlabel('Price ($)')
plt.ylabel('Number of Properties')
plt.show()
```
