**What does `density=True` do?** It rescales the histogram so the total area of the bars equals 1, which puts the histogram on the same scale as the normal curve.

## Example: heights of roller coasters
`roller_coaster.csv` is in this folder.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
import pandas as pd
```

```python
coaster = pd.read_csv('roller_coaster.csv')   # change: your file name
coaster = coaster.dropna()
heights = coaster['Max_Height']                # change: your column name
print(len(heights))
```

```python
plt.hist(heights, density=True)
mu, std = norm.fit(heights)
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'k', linewidth=2)
plt.title('Heights of Roller Coasters')        # change: your chart title
plt.xlabel('Height (ft)')                      # change: your axis label
plt.show()
```

```python
print('mean:', mu)
print('standard deviation:', std)
value = 100                                    # change: the value to check
print('z-score of', value, ':', (value - mu) / std)
```
