## Load the cleaned dataset
`dc_properties_clean.csv` is in this folder. It is the cleaned file from Lesson 2.6, provided here so you do not need your own copy.

```python
import pandas as pd
df_clean = pd.read_csv('dc_properties_clean.csv')
print(len(df_clean))
```

## 1. Average PRICE for every ward at once

```python
df_clean.groupby('WARD')['PRICE'].mean().round(2)
```

## 2. Mean and count together

```python
df_clean.groupby('WARD')['PRICE'].agg(['mean', 'count']).round(2)
```

## 3. Sort to find the highest and lowest

```python
ward_stats = df_clean.groupby('WARD')['PRICE'].agg(['mean', 'count']).round(2)
ward_stats.sort_values('mean', ascending=False)
```
