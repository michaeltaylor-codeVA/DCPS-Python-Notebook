## The sample PRICE list
These are the first ten sale prices from `DC_Properties_300_Student.csv` that have a recorded price.

```python
prices = [285000, 355000, 1111000, 1200000, 574900, 185000, 399900, 269000, 235000, 670000]
print(len(prices))
```

## 1. A basic for loop
The second line is indented because it is inside the loop. Jupyter Notebook adds the indent automatically when you press Enter after the colon.

```python
for price in prices:
    print(price)
```

## 2. Filtering with a conditional inside a loop

```python
for price in prices:
    if price > 500000:
        print(price)
```

```python
count = 0
for price in prices:
    if price > 500000:
        count = count + 1
print(count)
```

## 3. Trace this loop by hand first
On paper, predict every line this loop prints. Then run it and check.

```python
count = 0
for price in prices[:5]:
    if price > 500000:
        count = count + 1
    print(price, count)
```
