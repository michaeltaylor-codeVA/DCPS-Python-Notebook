**Finding the number of samples**: We can find the number of non-missing samples in one column using ```.count()```. 
* We have 146 rows if we only removed missing values in ```Top_Speed```
* We have 115 rows if we dropped all missing values

````python
speeds = coaster_df['Top_Speed']

print(speeds.count())
````

**Finding the max/min**: ```max()``` and ```min()``` can find the max and min of a column

````python
max_speed = speeds.max()
min_speed = speeds.min()

print('maximum:', max_speed)
print('minimum:', min_speed)
````

**Getting a Subset** Recall that we used subsets when we were creating visualizations. Suppose I want to only consider older roller coasters:

````python
old_coasters = coaster_df[coaster_df["Age_Group"] == "older"]
````

We can recalculate to learn more about our subset:

````python
old_speeds = old_coasters['Top_Speed']

print('number of samples:', old_speeds.count())
print('min:', old_speeds.min())
print('max:', old_speeds.max())
````

## Calculate Statistics

Pandas has a built-in method to calculate descriptive statistics of columns. Run this code block to see how it works.

````python
coaster_df.describe()
````

Let's calculate the descriptive statistics of the Top Speed on our own.

````python
speeds = coaster_df['Top_Speed']
````

Mean

````python
mean = speeds.mean()
print("The mean of top speed is", mean)
````

Median

````python
median = speeds.median()
print("The median top speed is", median)
````

Mode

````python
mode = speeds.mode()
print("The most common top speed is", mode)
````

Notice the output: `0    50.0`. The `0` on the left is the row label, not a second value — the mode is **50** mph. If two values were tied for most common, `.mode()` would list both, one per row. The `dtype` line tells you the mode is a number (```float64```).

Standard Deviation

````python
stdev = speeds.std()
print("The standard deviation of top speed is", stdev)
````

## Visualize

Remember that matplotlib has some helpful functions to visualize these statistics. For example, ```.boxplot```

````python
import matplotlib.pyplot as plt

plt.boxplot(speeds)
plt.show()
````

I could add to the boxplot function to visualize without the outliers too.

````python
plt.boxplot(speeds, showfliers=False)
plt.show()
````

Another common method for visualizing descriptive statistics is a histogram.

````python
plt.hist(speeds)
plt.show()
````

I could visualize the mean within the histogram using ```axvline(***.mean())```

````python
plt.hist(coaster_df['Top_Speed'])
plt.axvline(speeds.mean(), color = "red") #make it red so we can see it
plt.show()
````

## Final Code

If I were to put all of my code together, I would do it like this.



````python
import pandas as pd
import matplotlib.pyplot as plt

#Read and clean in data
coaster_df = pd.read_csv('roller_coaster.csv')
coaster_df.dropna(inplace = True, subset = ['Top_Speed'])
print(coaster_df.describe())
speeds = coaster_df['Top_Speed']

#calculate statistics
mean = speeds.mean()
median = speeds.median()
mode = speeds.mode()
stdev = speeds.std()

#print results
print("The mean top speed is", mean)
print("The median top speed is", median)
print("The most common top speed is", mode)
print("The standard deviation of top speed is", stdev)

#Create visualizations
plt.boxplot(speeds)
plt.show()

plt.hist(coaster_df['Top_Speed'])
plt.axvline(speeds.mean(), color = "red")
plt.show()

````
