## 2. Load the full Pokémon dataset and make two subsets

```python
pokemon_df = pd.read_csv('pokemon.csv')
legendary_df = pokemon_df[pokemon_df['is_legendary'] == 1]
non_legendary_df = pokemon_df[pokemon_df['is_legendary'] == 0]
print(len(pokemon_df), len(legendary_df), len(non_legendary_df))
```

## 3. Mean, median, and standard deviation of hp

```python
print(round(pokemon_df['hp'].mean(), 2), pokemon_df['hp'].median(), round(pokemon_df['hp'].std(), 2))
```

```python
print(round(legendary_df['hp'].mean(), 2), legendary_df['hp'].median(), round(legendary_df['hp'].std(), 2))
print(round(non_legendary_df['hp'].mean(), 2), non_legendary_df['hp'].median(), round(non_legendary_df['hp'].std(), 2))
```

## 4. Speed and the Empirical Rule

```python
mean = pokemon_df['speed'].mean()
sd = pokemon_df['speed'].std()
print(round(mean, 2), round(sd, 2))
```

```python
import matplotlib.pyplot as plt
plt.hist(pokemon_df['speed'])
plt.title('Pokémon Speed')
plt.xlabel('Speed')
plt.ylabel('Number of Pokémon')
plt.show()
```

```python
print(round(0.68 * len(pokemon_df)))
```

```python
within_1_sd = pokemon_df[(pokemon_df['speed'] >= mean - sd) & (pokemon_df['speed'] <= mean + sd)]
print(len(within_1_sd))
```
