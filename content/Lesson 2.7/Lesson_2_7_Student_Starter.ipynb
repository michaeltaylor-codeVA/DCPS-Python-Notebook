## Load the cleaned dataset
`dc_properties_clean.csv` is in this folder. It is the cleaned file from Lesson 2.6, provided here so you do not need your own copy.

```python
import pandas as pd
df_clean = pd.read_csv('dc_properties_clean.csv')
print(len(df_clean))
```

## 1. A dictionary of labeled counters

```python
ward_counts = {}
print(ward_counts)
```

## 2. Test the counting script on the first 10 values

```python
sample = df_clean['WARD'].head(10)
print(list(sample))
```

```python
ward_counts = {}
for ward in sample:
    if ward in ward_counts:
        ward_counts[ward] = ward_counts[ward] + 1
    else:
        ward_counts[ward] = 1
print(ward_counts)
```

## 3. Run the same script on the full WARD column

```python
ward_counts = {}
for ward in df_clean['WARD']:
    if ward in ward_counts:
        ward_counts[ward] = ward_counts[ward] + 1
    else:
        ward_counts[ward] = 1
print(ward_counts)
print(sum(ward_counts.values()))
```

## 4. Debugging practice: a script that runs but gives a wrong answer
This script should count the properties in Ward 2. It runs without an error, but the answer is wrong. Find and fix the bug.

```python
count = 0
for ward in df_clean['WARD']:
    if ward == 2:
        count = count + 1
print(count)
```
