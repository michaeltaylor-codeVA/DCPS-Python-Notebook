```python
import pandas as pd, numpy as np, matplotlib, scipy, sklearn
print('✅ pandas', pd.__version__, '| numpy', np.__version__, '| matplotlib', matplotlib.__version__)
print('✅ scipy', scipy.__version__, '| scikit-learn', sklearn.__version__)
```

```python
pokemon = pd.read_csv('Lesson 2.14/pokemon.csv')
print('✅ Read a dataset from the file browser:', pokemon.shape[0], 'rows,', pokemon.shape[1], 'columns')
```

```python
import matplotlib.pyplot as plt
plt.hist(pokemon['hp'], bins=20)
plt.title('Pokémon HP')
plt.savefig('device_check_chart.png')
plt.show()
import os
print('✅ Chart drawn and saved:', os.path.exists('device_check_chart.png'))
```

```python
from scipy.stats import linregress
from sklearn.linear_model import LogisticRegression
r = linregress(pokemon['attack'], pokemon['hp'])
model = LogisticRegression(max_iter=1000).fit(pokemon[['attack','defense']], pokemon['is_legendary'])
print('✅ Linear regression r =', round(r.rvalue, 3))
print('✅ Logistic regression accuracy =', round(model.score(pokemon[['attack','defense']], pokemon['is_legendary']), 3))
```
